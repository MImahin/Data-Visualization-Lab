## Creating DataFrame

### From Dictionary

```python
weather_data = {
 'day': ['1/1/2017','1/2/2017','1/3/2017','1/4/2017','1/5/2017','1/6/2017'],
 'temperature': [32,35,28,24,32,31],
 'windspeed': [6,7,2,7,4,2],
 'event': ['Rain', 'Sunny', 'Snow', 'Snow', 'Rain', 'Sunny']
}

df = pd.DataFrame(weather_data)
display(df)
```

# Matplotlib

Matplotlib is a versatile plotting library in Python. It allows the creation of static, animated, and interactive visualizations. The most commonly used module is pyplot.

## Import Matplotlib

```python
import matplotlib.pyplot as plt
```

## Basic Plotting

```python
# x = [1,3,2,4]
# y = [2,6,8,8]

x = [1,2,3,4]
y = [2,8,6,8]

plt.plot(x, y)
plt.title("Line Plot Example")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(False)
plt.show()
```

## Bar Plot

```python
categories = ["A", "B", "C"]
values = [4, 7, 1]

plt.bar(categories, values, color='skyblue')
plt.title("Bar Plot")
plt.show()
```

## Histogram

```python
import numpy as np
data = np.random.randn(1000)
# print(data)
plt.hist(data, bins=100, color='orange', edgecolor='black')
plt.title("Histogram of Normally Distributed Data")
plt.show()
```

## Scatter Plot

```python
x = np.random.randn(50)
y = np.random.randn(50)

plt.scatter(x, y, color='green')
plt.title("Scatter Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

## Customization and Subplots

```python
x = np.linspace(0, 10, 100)
print(f"x is :{x}")
y1 = np.sin(x)
#(x,y1)
print(f"y1 is {y1}")
y2 = np.cos(x)
print(f"y2 is {y2}")
#(x,y2)

plt.plot(x, y1, label='sin(x)')
plt.plot(x, y2, label='cos(x)')
plt.legend()
plt.title("Sine and Cosine")
plt.show()
```

```python
fig, axs = plt.subplots(2, 2, figsize=(10, 4))
axs[0][0].plot(x, y1)
axs[0][0].set_title("Sine")
axs[1][1].plot(x, y2)
axs[1][1].set_title("Cosine")
plt.show()
```

# Seaborn

Seaborn is built on top of Matplotlib and tightly integrates with Pandas. It provides a high-level interface for drawing attractive and informative statistical graphics.

## Import Seaborn

```python
import seaborn as sns
```

## Load Built-in Dataset

```python
tips = sns.load_dataset("tips")
tips.tail()
```

## Histogram

```python
sns.histplot(tips["total_bill"], kde=False)
plt.title("Distribution of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

## Scatter Plot

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="sex")
plt.title("Tip vs Total Bill by Gender")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

## Correlation Heatmap

```python
iris = sns.load_dataset("iris")
# print(iris.head())
corr = iris.corr(numeric_only=True)
# print(corr)
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap (Iris Dataset)")
plt.show()
```
