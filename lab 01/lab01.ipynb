
Task: Given two matrices A and B (of same dimensions), find C = 2A - 3B

```python
a = [[3, 4],
     [7, 8]]
b = [[1, 2],
     [5, 6]]

c = []
x=2
y=3
for i,j in zip(a,b):
    m=[]
    for k,l in zip(i,j):
        m.append(k*x-l*y)
    c.append(m)
print(c)
```

```python
d=[]
for i in range(len(a)):
    m=[]
    for j in range(len(a[i])):
        m.append(x*a[i][j]-y*b[i][j])
    d.append(m)
print(d)
```

```python
#Using Numpy Array
# condition is must be same dimension
a = np.array([[3, 4],
             [7, 8]])
b = np.array([[1, 2],
             [5, 6]])
c=2*a-3*b
print(c)
```
